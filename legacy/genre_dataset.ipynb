```python
def get_columns(path):
    with open(path, "r") as file:
        header = file.readline().strip()
        columns = header.split(",")
    return columns
```

```python
def match_columns(columns, dataset):
    """Match the given columns with the dataset.
    Args:
        columns (list): List of columns to match.
        dataset (list): List of columns in the dataset.
    Returns:
        list: List of matched columns.
    """
    matched_columns = []
    for column in columns:
        if column in dataset:
            matched_columns.append(column)
    return matched_columns
```

```python
def check_duplicate_rows(dataset_path):
    # Load the dataset
    dataset = open(dataset_path, "r").readlines()
    # Remove the header
    dataset = dataset[1:]  # Skip the header line
    # Create a set to store unique rows
    unique_rows = set()
    duplicate_rows = []
    for row in dataset:
        row = row.strip()
        if row in unique_rows:
            duplicate_rows.append(row)
        else:
            unique_rows.add(row)
```
